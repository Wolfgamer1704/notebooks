## 2. Der Black-Box-Test

### Erklärung

**Ziel (Folie 5):** Überprüfen, ob ein Softwaresystem mit seiner **Spezifikation** übereinstimmt.
- beschränkt sich auf **funktionsorientiertes** Testen (Was kommt bei welcher Eingabe heraus?)
- wird **anhand der Spezifikation** erstellt – die Spezifikation ist das **Pflichtenheft**
- wird **nicht** vom Entwickler der zu testenden Software erstellt
- Das Programm ist eine „Black Box“: Man sieht nur **Eingabe (In)** und **Ausgabe (Out)**, nicht den Code dazwischen.

Ein **Test** führt das Programm mit festgelegten Eingaben aus und vergleicht die **tatsächliche Ausgabe** mit der **erwarteten Ausgabe**. Die erwarteten Werte kommen immer aus der Spezifikation, nie aus dem Programm selbst – sonst schreibt man nur ab, was das Programm sowieso macht.

**Vorteile (Folie 6):**
- verhindert, dass Programmierer Tests „um ihre eigenen Fehler herum“ entwickeln
- deckt eine **unvollständige Spezifikation** auf, weil beim Erstellen der Tests oft Fragen auftreten

**Nachteile (Folie 6):**
- keine Garantie für Fehlerfreiheit
- die Spezifikation deckt spätere Detail- und Implementierungsentscheidungen nicht ab

**Wer testet? (Folie 9):** Anwender oder ein anderes Team als die Entwickler. Unklare Fälle **mit dem Auftraggeber besprechen** – so werden Missverständnisse aufgedeckt.

## 4. Testfälle auswählen – Äquivalenzklassen

### Erklärung

**Folie 7 – Auswahl der Testfälle.** Wichtig sind:
- jede mögliche **Kombination** an Eingaben – **repräsentativ, nicht vollständig**
- **Grenzwerte**

Die Folie nennt als Theorie dahinter die **Äquivalenzklassenmethode** (daneben: Klassifikationsbaum, Entscheidungstabellen, zustandsbezogene Tests, Use-Case-Tests).

**Äquivalenzklassen:** Man teilt jeden Eingabebereich in Gruppen auf, in denen sich das Programm **gleich verhält**. Aus jeder Klasse reicht **ein Vertreter (Repräsentant)**. So testet man „repräsentativ, nicht vollständig“.
- **gültige Klassen:** laut Spezifikation erlaubte Eingaben
- **ungültige Klassen:** nicht erlaubte Eingaben (negative Zahlen, falsche Buchstaben, Text statt Zahl)

*Beispiel an einem anderen Fall (Kinoeintritt nach Alter):*

| Klasse | Bereich | Verhalten | Repräsentant |
|---|---|---|---|
| ungültig | Alter < 0 | Fehlermeldung | −3 |
| gültig | 0 bis 5 | frei | 4 |
| gültig | 6 bis 17 | Kinderpreis | 12 |
| gültig | ab 18 | Erwachsenenpreis | 30 |
| ungültig | keine Zahl | Fehlermeldung | "abc" |

Hängt das Verhalten von **mehreren Eingaben gemeinsam** ab, müssen die **Kombinationen** betrachtet werden.

## 5. Grenzwerte und spezielle Werte

### Erklärung

**Grenzwerte:** Die meisten Fehler passieren an den Rändern der Klassen: `<` statt `<=`, `>` statt `>=`, eine Grenze um eins verschoben. Darum testet man **genau auf der Grenze und direkt daneben**.

Das Beispiel aus Folie 9 zeigt genau das: Getestet werden **499** („zu geringe Menge“) und **500** („Untergrenze Rabatt“) – also die letzte Zahl ohne und die erste Zahl mit Rabatt.

**Spezielle Werte** (Folie 7) sind Eingaben, bei denen Programme gern stolpern, auch wenn sie in keine „normale“ Klasse passen (Ergänzung, typische Beispiele):
- 0, negative Zahlen
- sehr große Zahlen
- leere Eingabe (einfach Enter drücken)
- Text statt Zahl
- Großbuchstaben statt Kleinbuchstaben (`W` statt `w`)
- mehrere Zeichen statt einem (`weich` statt `w`)

*Beispiel an einem anderen Fall (Kino, Erwachsenenpreis ab 18):* Teste 17 und 18. Steht im Code `alter > 18` statt `alter >= 18`, fällt das **nur** beim Test mit 18 auf.

### Erklärung

**Folie 15:** Der Grey-Box-Test **verbindet Black-Box- und White-Box-Test**.
- Der Test wird z. B. **vor der Implementierung** erstellt (aus der Spezifikation) **oder**
- von einem **Test-Team aus Entwicklern, das Zugang zum Quellcode hat**.
- Er wird **kontinuierlich weiterentwickelt**, wenn sich während des Entwurfs etwas ändert.
- Er erfordert **hohe Disziplin**.
- Er ist in der Regel Bestandteil **agiler Prozesse**.

## 8. Testbar programmieren – Berechnung von Ein- und Ausgabe trennen (Ergänzung)

### Erklärung

Im Moodle-Programm passiert alles in `main`: Einlesen, Rechnen, Ausgeben. So etwas kann man nur von Hand testen – bei jedem Testfall muss jemand tippen und ablesen.

Testbar wird es, wenn man trennt:
- **Berechnungsmethode:** bekommt alle Werte als **Parameter**, liefert das Ergebnis mit **`return`**, kein Einlesen, kein `println`.

*Beispiel an einem anderen Fall (Kinoticket):*

```java
// gut testbar: Parameter rein, Ergebnis raus
double ticketpreis(int alter) {
    if (alter >= 18) return 10.0;
    return 6.0;
}

// Test ohne Tippen:
System.out.println(ticketpreis(17));   // erwartet 6.0
System.out.println(ticketpreis(18));   // erwartet 10.0
```

## 9. Automatisierte Tests und ungültige Eingaben (Ergänzung)

### Erklärung

Statt jedes Ergebnis von Hand zu vergleichen, schreibt man eine **Testmethode**: Sie ruft die zu testende Methode auf, vergleicht Ist mit Soll und meldet „bestanden“ oder „Abweichung“.

**Achtung bei Kommazahlen:** `double`-Werte nie mit `==` vergleichen. Durch Rundungsfehler ist z. B. `0.1 + 0.2 == 0.3` **false** (Ergebnis: 0.30000000000000004). Stattdessen mit Toleranz:

```java
boolean fastGleich(double a, double b) {
    return Math.abs(a - b) < 0.001;
}
```

**Ungültige Eingaben:** Eine reine Berechnungsmethode kann nicht „nochmal fragen“. Üblich ist, eine **Exception zu werfen** (z. B. `IllegalArgumentException`); das Nachfragen übernimmt `main`. Ein Test für einen Fehlerfall ist **bestanden, wenn die Exception kommt**.

*Beispiel an einem anderen Fall:*

```java
int bestanden = 0, gesamt = 0;

    double ist = ticketpreis(alter);
    boolean ok = Math.abs(ist - soll) < 0.001;
    gesamt++;
    if (ok) bestanden++;
    System.out.println("| " + alter + " | " + soll + " | " + ist + " | " + (ok ? "bestanden" : "Abweichung") + " |");
}

pruefeTicket(17, 6.0);
pruefeTicket(18, 10.0);

try {
    Integer.parseInt("abc");
    System.out.println("Abweichung – keine Exception");
} catch (NumberFormatException e) {
    System.out.println("bestanden – Exception: " + e.getMessage());
}

System.out.println(bestanden + " von " + gesamt + " bestanden");
```

So lässt sich auch unterscheiden, was du in deiner `testing()`-Methode notiert hast: ob das Programm einen Fehler **sauber meldet** oder **unkontrolliert abstürzt**.

## 10. Der White-Box-Test – Überdeckungsarten

### Erklärung

**Ziel (Folie 10):** Sicherstellen, dass die Testfälle den **Quellcode ausreichend überdecken**. Man kennt die innere Funktionsweise – es wird **am Code** geprüft.

**Überdeckungsarten (Folie 11 und 12):**

| Art | Was muss erreicht werden? |
|---|---|
| **Anweisungsüberdeckung** | Jede **Anweisung** wird mindestens einmal ausgeführt. |
| **Zweigüberdeckung** (Kantenüberdeckung) | Jeder mögliche **Zweig** des Kontrollflusses wird durchlaufen. |
| **Bedingungsüberdeckung** | Alle **ausschlaggebenden Belegungen** der Teilbedingungen in logischen Ausdrücken werden durchlaufen. |
| **Pfadüberdeckung** | Alle **Kombinationen von Zweigen** (alle Wege durch das Programm) werden geprüft. |

**Zweigüberdeckung (Folie 11):**
```java
if ( ... )
    anweisungA;
else
    anweisungB;
// Test 1: zum Ausführen von A
// Test 2: zum Ausführen von B
```
Auch ein `if` **ohne** `else` hat zwei Zweige: „wird ausgeführt“ und „wird übersprungen“. Für Anweisungsüberdeckung reicht dort der erste Fall, für Zweigüberdeckung braucht man beide.

**Bedingungsüberdeckung bei `if (a && b)` (Folie 11):**
Theoretisch gibt es 4 Belegungen (tt, tf, ft, ff). **3 Fälle reichen** aber: Wenn `a` false ist, wird `b` **nicht mehr geprüft** (Kurzschlussauswertung von `&&`). Die Fälle „a = false, b = true“ und „a = false, b = false“ verhalten sich also gleich.

**Zweig- vs. Pfadüberdeckung (Folie 12):** Zwei `if` hintereinander:
- Zweigüberdeckung: 2 Tests reichen (beide J, beide N) – jeder Zweig war einmal dran.
- Pfadüberdeckung: 4 Tests (J-J, J-N, N-J, N-N) – jede Kombination.
Bei vielen Verzweigungen wächst die Zahl der Pfade sehr schnell (Verdopplung pro `if`).

**Anmerkung zu Folie 11:** Zeilenüberdeckung ist nicht dasselbe wie Anweisungsüberdeckung. Stehen zwei Anweisungen in einer Zeile (z. B. `if (x) y++;`), reicht für die Zeilenüberdeckung schon, dass die Bedingung geprüft wurde – auch wenn `y++` nie ausgeführt wurde.

## 11. Black-Box und White-Box im Vergleich

### Erklärung

**White-Box – Vorteile (Folie 13):**
- findet **toten Code** (Code, der nie ausgeführt wird)
- eignet sich gut, um bereits aufgetretene Fehler zu **lokalisieren**

**White-Box – Nachteile (Folie 13):**
- testet **nicht, ob das Programm seinen Zweck erfüllt**
- **fehlender** Quellcode wird nicht entdeckt
- Entwickler neigen dazu, um die eigenen Fehler herum zu testen
- kein Maßstab für effizienten Code und Lesbarkeit (z. B. Spaghetti-Code)

**Vergleich (Folie 14):**

| | Black-Box | White-Box |
|---|---|---|
| **Pro** | testet die Spezifikation · Testsequenzen unabhängig von der Plattform | Komponententests möglich · toter Code wird aufgedeckt · Automatisierung durch Tool-Unterstützung |
| **Contra** | zwei Fehler können sich gegenseitig aufheben und bleiben unentdeckt · größerer organisatorischer Aufwand (Fachteam nötig) · nachträglich eingefügte Funktionalität wird nicht berücksichtigt · abhängig von guter Spezifikation | Testen um Fehler herum · fehlender Code wird nicht gefunden · Spaghetti-Code, schlechte Lesbarkeit, Abweichung vom Style-Guide können nur von Dritten gefunden werden |

Die Unterscheidung hängt teilweise von der **Perspektive** ab: Der Test einer Komponente ist aus Sicht des Gesamtsystems ein White-Box-Test, aus Sicht der Komponente selbst kann er ein Black-Box-Test sein.

## 12. Der Debugger

### Erklärung

**Fehlerarten (Folie 20):**
- **Syntaxfehler:** Das Programm compiliert nicht – der Compiler zeigt die Stelle an.
- **Logische Fehler:** Das Programm compiliert, arbeitet aber falsch. **Der Debugger hilft, logische Fehler zu finden.**

**Werkzeuge (Folie 19–25):**
- **Haltepunkt (Breakpoint):** Dort hält das Programm beim Debuggen an, damit man sich die Inhalte der Variablen ansehen kann.
- **Variables-Fenster (Watch-Variablen):** zeigt alle aktuellen Variablen mit ihren Werten.
- **Steppen:** bis zum nächsten Haltepunkt weitermachen · anhalten · **in die Methode hineinspringen** · **einzelne Anweisung ausführen**
- **Expressions:** eigene Ausdrücke beobachten (über das Kontextmenü, rechte Maustaste), z. B. `menge * 2.5`

Die Folien zeigen Eclipse (Debug-Perspektive, Haltepunkt per Doppelklick links neben die Zeilennummer). **In VS Code (Ergänzung):**

| einzelne Anweisung ausführen (Step Over) | **F10** |
| in die Methode hineinspringen (Step Into) | **F11** |
| aus der Methode heraus (Step Out) | **Umschalt+F11** |
| anhalten | **Umschalt+F5** |

## Zusammenfassung – meine wichtigsten Punkte für die Klausur

- **Test** = Ist mit Soll vergleichen. Das Soll kommt **immer aus der Spezifikation (Pflichtenheft)**, nie aus dem Programm.

- **Fehlerkosten** steigen mit jeder Phase: Entwurf → Konstruktion → Abnahme → Betrieb.
- **Black-Box:** nur Spezifikation, nicht vom Entwickler. Findet fehlende Funktionen und Lücken in der Spezifikation. Keine Garantie, abhängig von guter Spezifikation.
- **White-Box:** am Code, Überdeckung. Findet toten Code, hilft beim Lokalisieren. Findet **keinen fehlenden Code**, prüft nicht den Zweck.
- **Grey-Box:** beides zusammen, z. B. eigener Code oder Test vor der Implementierung; agile Prozesse.
- **Testfälle auswählen:** Äquivalenzklassen (gültig und ungültig, ein Repräsentant je Klasse), alle Kombinationen, **Grenzwerte (beide Seiten!)**, spezielle Werte (0, negativ, leer, Text).
- **Protokoll (Folie 9):** Was? | Eingaben | Voraussetzungen | erwartetes Ergebnis | Bestanden oder Abweichung. Soll vorher von Hand rechnen.
- **Überdeckung:** Anweisung (jede Anweisung) < Zweig (jede Bedingung J und N, auch `if` ohne `else`) < Pfad (jede Kombination). Bedingungsüberdeckung: bei `a && b` reichen 3 Fälle wegen Kurzschluss.
- **Debugger:** findet **logische** Fehler (Syntaxfehler findet der Compiler). Haltepunkt, Variables/Watch, Step Over (F10), Step Into (F11), weiter (F5).
